```python
def write_json(german_words, english_words):
    with open('./data/A1_Nouns.json', 'r+') as file:
        file_data = json.load(file)
        print(file_data)
        for word_index in range(len(german_words)):
            file_data['words'].append({"german": german_words[word_index], "english": english_words[word_index]})
        file.seek(0)
        json.dump(file_data, file, indent=4)
```

```python
write_json(german_words=german_words, english_words=english_words)
```
